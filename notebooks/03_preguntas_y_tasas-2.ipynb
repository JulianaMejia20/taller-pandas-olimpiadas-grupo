# Preguntas y tasas de medalla (Puntos 9 y 10)
Se cargan los datos igual que en el notebook 01.

```python
import pandas as pd
import numpy as np

df_atletas = pd.read_csv('../data/athlete_events.csv')
df_regiones = pd.read_csv('../data/noc_regions.csv')

df_atletas['IMC'] = df_atletas['Weight'] / ((df_atletas['Height'] / 100) ** 2)
df_atletas = pd.merge(df_atletas, df_regiones, on='NOC', how='left')
df_atletas['medalla'] = df_atletas['Medal'].notna().astype(int)

print(df_atletas.shape)
```

### 9. Preguntas puntuales

```python
print("a) Edad promedio:", round(df_atletas['Age'].mean(), 2))

print("\nb) Registros con y sin medalla:")
print(df_atletas['medalla'].value_counts().rename({0: 'Sin medalla', 1: 'Con medalla'}))

print("\nc) Estatura promedio de los que ganaron oro:",
      round(df_atletas.loc[df_atletas['Medal'] == 'Gold', 'Height'].mean(), 2))

print("\nd) Ediciones distintas:", df_atletas['Games'].nunique())
print("   Países (NOC) distintos:", df_atletas['NOC'].nunique())

print("\ne) Edad mínima:", df_atletas['Age'].min(), " Edad máxima:", df_atletas['Age'].max())

print("\nf) Registros por temporada:")
print(df_atletas['Season'].value_counts())

ediciones = df_atletas.groupby('ID')['Games'].nunique()
print("\ng) Atletas en una sola edición:", (ediciones == 1).sum())
print("   Atletas en más de una edición:", (ediciones > 1).sum())
```

### Respuestas

* **a)** La edad promedio de los atletas es de 25.56 años.
* **b)** 39,783 registros terminaron con medalla y 231,333 sin medalla, o sea que solo cerca del 14.7% de las participaciones terminan en medalla.
* **c)** La estatura promedio de los atletas que ganaron oro es de 177.9 cm, más alta que el promedio general de todos los atletas (175.3 cm). Se calcula solo con los que tienen estatura registrada.
* **d)** Hay 51 ediciones y 230 códigos NOC distintos. Ojo que un mismo país puede tener varios códigos a lo largo de la historia.
* **e)** La edad mínima es 10 años y la máxima 97. La de 97 llama la atención y seguramente corresponde a las competencias de arte, que antes eran parte de los Juegos.
* **f)** 222,552 registros son de Verano y 48,564 de Invierno.
* **g)** 98,450 atletas participaron en una sola edición y 37,121 en más de una, o sea que cerca del 73% de los atletas solo fue a unos Juegos.

### 10. Hipótesis y verificación con tasas

```python
def resumen(col):
    t = df_atletas.groupby(col)['medalla'].agg(medallas='sum', registros='count')
    t['tasa_%'] = (t['medallas'] / t['registros'] * 100).round(2)
    return t

# a) mujeres vs hombres
display(resumen('Sex'))

# b) menores de 18 vs adultos (sin contar los que no tienen edad)
edad = df_atletas.dropna(subset=['Age']).copy()
edad['menor_18'] = np.where(edad['Age'] < 18, 'Menor de 18', '18 o más')
t_b = edad.groupby('menor_18')['medalla'].agg(medallas='sum', registros='count')
t_b['tasa_%'] = (t_b['medallas'] / t_b['registros'] * 100).round(2)
display(t_b)

# c) menores de 18 o 35+ vs el resto
edad['grupo'] = np.where((edad['Age'] < 18) | (edad['Age'] >= 35), 'Menor de 18 o 35+', '18 a 34')
t_c = edad.groupby('grupo')['medalla'].agg(medallas='sum', registros='count')
t_c['tasa_%'] = (t_c['medallas'] / t_c['registros'] * 100).round(2)
display(t_c)

# d) USA vs otros paises
df_atletas['es_usa'] = np.where(df_atletas['NOC'] == 'USA', 'USA', 'Otros países')
display(resumen('es_usa'))

# e) verano vs invierno
display(resumen('Season'))

resumen('Sex').to_csv('../outputs/tablas/medallas_por_sexo.csv')
resumen('Season').to_csv('../outputs/tablas/medallas_por_temporada.csv')
```

### Hipótesis y resultados

* **a) Mujeres vs hombres.** Hipótesis: los hombres ganaron más medallas porque hay muchos más hombres en los datos. En conteo se confirma (28,530 de hombres contra 11,253 de mujeres), pero proporcionalmente pasa lo contrario: la tasa de medalla de las mujeres (15.10%) es un poco mayor que la de los hombres (14.51%). La diferencia en conteo se explica porque hay muchos más registros de hombres.
* **b) Menores de 18 vs adultos.** Hipótesis: los adultos ganan más medallas en conteo y también en proporción. Se confirma: los adultos tienen 37,872 medallas contra 1,179 de los menores, y su tasa también es más alta (15.20% contra 9.43%).
* **c) Menores de 18 o 35+ vs el resto.** Hipótesis: el grupo de 18 a 34 gana más porque es la edad de mejor rendimiento. Se confirma en conteo (35,066 contra 3,985) y en proporción (15.27% contra 12.44%).
* **d) USA vs otros países.** Hipótesis: en conteo ganan los otros países porque son muchos más, pero proporcionalmente USA tiene una tasa más alta. Se confirma: los otros países suman 34,146 medallas contra 5,637 de USA, pero la tasa de USA (29.90%) es más del doble que la del resto (13.54%).
* **e) Verano vs Invierno.** Hipótesis: en Verano se ganan más medallas porque hay muchas más pruebas. Se confirma en conteo (34,088 contra 5,695) y también en proporción (15.32% contra 11.73%).
* **Conclusión:** comparar solo conteos puede engañar, porque los grupos tienen tamaños muy distintos. El caso más claro es el de mujeres y hombres: en conteo ganan los hombres, pero en tasa las mujeres quedan un poco arriba.
